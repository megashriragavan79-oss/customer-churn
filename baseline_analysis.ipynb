# Customer Churn: Data Profiling & Non-ML Baseline
This notebook inspects the supplied dataset and establishes a majority-class baseline. It intentionally does not train a predictive model: the dataset contains only 12 records.


```python
import pandas as pd
from pathlib import Path

# Update this path if running outside the supplied environment.
DATA_PATH = Path('customer-churn-training.csv')
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
display(df.info())
display(df.isna().sum().rename('missing_values').to_frame())
display(df.describe(include='all').T)
```

```python
target = 'churned'
print('Target counts:')
display(df[target].value_counts(dropna=False).sort_index())
print('Target proportions:')
display(df[target].value_counts(normalize=True, dropna=False).sort_index())
assert set(df[target].dropna().unique()).issubset({0, 1}), 'Expected binary target values 0/1'
```

## Non-ML baseline: majority class
The baseline predicts the most common label for every row. This is a benchmark only. Do not interpret training-set metrics as out-of-sample performance.


```python
majority_class = int(df[target].mode().iloc[0])
baseline_predictions = pd.Series(majority_class, index=df.index, name='baseline_prediction')
baseline_accuracy = (baseline_predictions == df[target]).mean()
print('Majority class:', majority_class)
print('In-sample baseline accuracy (descriptive only):', round(float(baseline_accuracy), 4))
print('Confusion table:')
display(pd.crosstab(df[target], baseline_predictions, rownames=['actual'], colnames=['predicted']))
```
